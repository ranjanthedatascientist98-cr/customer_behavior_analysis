We can see 37 null values in "Review Rating" columns . So we are going to fill it with median values with respective category using groupby . 


```python
df["review_rating"]=df.groupby('category')["review_rating"].transform(lambda x : x.fillna(x.median()))
```

```python
# rechecking the missing values 
df.isnull().sum()
```

- Now we can see that some column names have space , Upper case which may be problemetic during sql and writing names . So we will convert them into sneak casing (combination of lower case & "_") . 
- We are going to us .str beacuse 
   - df.columns is NOT a single string
   - .columns = collection of strings
   - .lower() works only on one string at a time
   - .str is a vectorized string accessor
   - .strip() => Removes extra spaces from beginning and end

```python
df.columns = df.columns.str.lower().str.replace(" ", "_").str.strip()

# renaming the column name 
df.rename(columns={'purchase_amount_(usd)' : 'purchase_amount'},inplace =True)
```

```python
# rechecking the restructered column name  
df.columns
```

- To understand the market and purchasing behaviour  we are coenverting the 'age'  as 'age_groups' .
- qcut() => Equal number of rows per bin
- Syntax : pd.qcut(data, q)

```python
labels = ['Young-Adult','Adult','Middle-Aged','Seniors']
df["Age_groups"]=pd.qcut(df["age"],q=4,labels=labels)
df[['age','Age_groups']].tail(10)
```

```python
unique=df['frequency_of_purchases'].value_counts()
unique
```

```python
frequency_maping = {
 'Fortnightly'    : 14,
 'Every 3 Months' : 90,
 'Annually'       :365,
 'Quarterly'      :14,
 'Monthly'        :30,
 'Bi-Weekly'      :14, 
 'Weekly'         :7    
}

# create new column frequency_of_purchases_days
df["frequency_of_purchases_days"] = df["frequency_of_purchases"].map(frequency_maping)
df[["frequency_of_purchases_days","frequency_of_purchases"]]
```
